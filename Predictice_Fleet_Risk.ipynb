```python
import pandas as pd

def analyze_transporter_performance(units_a, units_b, units_c, units_d):
    """
    Calculates CGPA and Risk Status based on the Jubilee Support 'CheckIT' Logic.
    """
    # Mapping based on IHS CheckIT.pdf
    ratings = {
        'A': {'score': 4, 'units': units_a},
        'B': {'score': 3, 'units': units_b},
        'C': {'score': 2, 'units': units_c},
        'D': {'score': 1, 'units': units_d}
    }

    total_units = sum(item['units'] for item in ratings.values())
    total_weighted_points = sum(item['score'] * item['units'] for item in ratings.values())

    cgpa = total_weighted_points / total_units

    # Risk Assessment Logic
    if cgpa < 2.0:
        status = "WATCHLIST (Critical Risk)"
    elif 2.0 <= cgpa <= 2.5:
        status = "SWEET SPOT (Acceptable)"
    else:
        status = "COMMENDABLE (Low Risk)"

    return round(cgpa, 2), status, total_units

# Example: Transporter with 40 units (based on IHS CheckIT sample)
score, risk_status, fleet_size = analyze_transporter_performance(20, 13, 5, 2)

print(f"Analysis for Fleet Size: {fleet_size}")
print(f"Transporter CGPA: {score}")
print(f"Risk Category: {risk_status}")
```
